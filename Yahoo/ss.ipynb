```python
import pandas as pd

# Load the uploaded CSV file
file_path = 'financial_data.csv'
financial_df = pd.read_csv(file_path)

# Display the first few rows of the DataFrame
financial_df.head()
```

```python
# Sort DataFrame by a specific column in descending order (e.g., 'Market Cap')
sorted_df = financial_df.sort_values(by='ROE', ascending=False)

# Save the sorted DataFrame to a new CSV file
sorted_df.to_csv('RoE.csv', index=False)
print("Sorted data saved to RoE.csv")
```

```python


# Sort DataFrame by a specific column in ascending order (e.g., 'Market Cap')
sorted_df = financial_df.sort_values(by='P/E Ratio', ascending=True)

# Save the sorted DataFrame to a new CSV file
sorted_df.to_csv('value.csv', index=False)
print("valur.csv")
```
