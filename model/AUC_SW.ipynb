```python
#import module
import pandas as pd
from sklearn.metrics import roc_auc_score
```

**CMA**

```python
#Read .csv file
data = pd.read_csv('model_results.csv')

#chose colum
prob = data['prediction_probabilities']
y_true = data['actual_test_values']

#calculate AUC
auc = roc_auc_score(y_true, prob)

print("CMA AUC =", auc)
```
